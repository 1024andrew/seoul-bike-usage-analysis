```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False
```

```python
hourly_usage = pd.read_csv(
    r"C:\Users\1024a\Desktop\portfolior\mysql_projects\bike_analysis\data\hourly_usage.csv"
)

hourly_usage.head()
```

```python
#시간대별 이용량 그래프 그리기
plt.figure(figsize=(10, 5))

plt.plot(
    hourly_usage["rental_hour"],
    hourly_usage["total_usage"],
    marker="o"
)

plt.title("Hourly Public Bike Usage")
plt.xlabel("Rental Hour")
plt.ylabel("Total Usage")
plt.xticks(range(0, 24))
plt.grid(True)

plt.show()
```

```python
#대여소별 이용량 top 10
top10_station_usage = pd.read_csv(
    r"C:\Users\1024a\Desktop\portfolior\mysql_projects\bike_analysis\data\top10_station_usage.csv"
)

top10_station_usage.head()
```
